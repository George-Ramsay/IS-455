# Part 1: Load and Explore the Dataset

In this section, you'll load the dataset and perform initial exploration to understand its structure and characteristics.

## Question 1: Load the Dataset

Load the `housing.csv` file into a Pandas DataFrame named `df`. Display the first few rows and the dataset shape.

**Checkpoint Question**: What is the total number of rows in the dataset?

```python
# Question 1:
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.preprocessing import MinMaxScaler
from pathlib import Path
import sys

# Make ML-Pipeline-Kit importable regardless of notebook working directory
library_candidates = [Path.cwd() / 'ML-Pipeline-Kit', Path.cwd().parent / 'ML-Pipeline-Kit']
for candidate in library_candidates:
    if candidate.exists():
        sys.path.insert(0, str(candidate.resolve()))
        break

from ml_library import missing_data_diagnostics, missing_data_clean

# Load housing.csv from common execution locations
data_candidates = [
    Path('housing.csv'),
    Path('Assignment-3') / 'housing.csv',
    Path.cwd().parent / 'Assignment-3' / 'housing.csv',
]

for path in data_candidates:
    if path.exists():
        data_path = path
        break
else:
    raise FileNotFoundError('Could not locate housing.csv in expected locations.')

df = pd.read_csv(data_path)
display(df.head())
print(f'Dataset shape: {df.shape}')
print(f'Total rows: {df.shape[0]}')

```

# Part 2: Data Cleaning

In this section, you'll clean the dataset using techniques from prior chapters. Focus on handling missing values and preparing the data for modeling.

## Question 4: Remove Rows with Missing Target Variable

Since we cannot use rows where the target variable (`SalePrice`) is missing, drop all rows where `SalePrice` is null. This is critical because we need the label to build and evaluate our model.

# Part 3: Prepare Features for Multiple Linear Regression

In this section, you'll prepare your features for MLR modeling. This includes separating the label from features, dummy coding categorical variables, and optionally scaling numeric features.

## Question 6: Separate Label and Features

Create two separate variables:
- `y`: The target variable (`SalePrice`)
- `X`: All other columns as features (excluding `SalePrice`)

Display the shape of both `y` and `X` to confirm they are set up correctly.

**Checkpoint Question**: How many features (columns) are in your feature matrix `X` before dummy coding?

```python
# Question 6:
y = df['SalePrice'].copy()
X = df.drop(columns=['SalePrice']).copy()

print(f'y shape: {y.shape}')
print(f'X shape: {X.shape}')
print(f'Number of features before dummy coding: {X.shape[1]}')

```

## Question 7: Dummy Code Categorical Variables

Convert all categorical (object type) columns in `X` to dummy variables using `pd.get_dummies()`. Remember to use `drop_first=True` to avoid perfect multicollinearity.

After dummy coding, display the new shape of `X` and show the first few rows.

**Checkpoint Question**: After dummy coding, how many features (columns) are in your feature matrix `X`?

```python
# Question 7:
X = pd.get_dummies(X, drop_first=True)

print(f'X shape after dummy coding: {X.shape}')
display(X.head())
print(f'Number of features after dummy coding: {X.shape[1]}')

```

## Question 8: Ensure All Features are Numeric

Verify that all columns in `X` are now numeric. Convert any remaining boolean or object columns to numeric if necessary. This ensures compatibility with statsmodels.

**Checkpoint Question**: How many non-numeric columns remain in `X` after dummy coding and conversion? (The answer should be 0)

```python
# Question 8:
# Convert any non-numeric columns (typically bool dummies) to integers
non_numeric_before = X.select_dtypes(exclude=[np.number]).columns.tolist()
if non_numeric_before:
    X[non_numeric_before] = X[non_numeric_before].astype(int)

non_numeric_after = X.select_dtypes(exclude=[np.number]).columns.tolist()

print(f'Non-numeric columns before conversion: {len(non_numeric_before)}')
print(f'Non-numeric columns after conversion: {len(non_numeric_after)}')

```

# Part 4: Build Multiple Linear Regression Model

In this section, you'll build your MLR model using statsmodels and interpret the results.

## Question 9: Build the MLR Model

Using `statsmodels.api.OLS`, build a multiple linear regression model to predict `SalePrice` from all features in `X`. Remember to:
1. Add a constant term to your feature matrix (use `.assign(const=1)` or `sm.add_constant()`)
2. Fit the model
3. Display the model summary

**Checkpoint Question**: What is the R-squared value for your model? Round your answer to 4 decimal places.

```python
# Question 9:
X_const = sm.add_constant(X, has_constant='add')
model = sm.OLS(y, X_const)
results = model.fit()

print(results.summary())
print(f'R-squared: {results.rsquared:.4f}')

```

## Question 10: Extract Model Fit Statistics

Extract and display the following model fit statistics from your regression results:
- R-squared
- Adjusted R-squared
- Number of observations
- Number of features (excluding the constant)

**Checkpoint Question**: What is the difference between R-squared and Adjusted R-squared value? Round your answer to 5 decimal places.

```python
# Question 10:
model_stats = pd.Series({
    'R-squared': results.rsquared,
    'Adjusted R-squared': results.rsquared_adj,
    'Number of observations': int(results.nobs),
    'Number of features (excluding constant)': int(X_const.shape[1] - 1),
})

display(model_stats)
r2_difference = results.rsquared - results.rsquared_adj
print(f'R-squared - Adjusted R-squared: {r2_difference:.5f}')

```

## Question 11: Identify Statistically Significant Features

From your model results, identify features that are statistically significant. Print a summary of only those features showing:
1. The feature name
2. The coefficient value
3. The p-value

Sort this list by p-value (smallest first) to see the most significant features.

**Checkpoint Question**: How many features have statistically significant p-values? (Exclude the constant term from your count)

```python
# Question 11:
pvals = results.pvalues.drop(labels='const', errors='ignore')
sig_index = pvals[pvals < 0.05].index

significant_features = pd.DataFrame({
    'feature': sig_index,
    'coefficient': results.params[sig_index].values,
    'p_value': pvals[sig_index].values,
}).sort_values('p_value', ascending=True)

display(significant_features)
print(f'Number of statistically significant features (excluding constant): {significant_features.shape[0]}')

```

## Question 12: Find the Feature with Largest Coefficient

Identify which feature (excluding the constant) has the largest absolute coefficient value. This helps understand which feature has the strongest association with SalePrice in your model.

**Checkpoint Question**: What is the name of the feature with the largest absolute coefficient value? (Exclude the constant term)

```python
# Question 12:
coef_no_const = results.params.drop(labels='const', errors='ignore')
largest_feature = coef_no_const.abs().idxmax()
largest_coef = coef_no_const[largest_feature]

print(f'Feature with largest absolute coefficient: {largest_feature}')
print(f'Coefficient value: {largest_coef:.4f}')

```

## Question 13: Calculate In-Sample Prediction Metrics

Calculate and display the following in-sample prediction metrics:
1. **Mean Absolute Error (MAE)**: The average absolute difference between actual and predicted values
2. **Root Mean Squared Error (RMSE)**: The square root of the average squared prediction error

Use the fitted values from your model (`results.fittedvalues`) to compute these metrics.

**Checkpoint Question**: What is the RMSE value? Round your answer to 2 decimal places.

```python
# Question 13:
y_pred = results.fittedvalues
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print(f'MAE: {mae:.2f}')
print(f'RMSE: {rmse:.2f}')

```

## Question 14: Interpret a Categorical Feature Coefficient

Interpret the coefficient for the `CentralAir_Y` dummy variable in your model. This variable indicates whether a home has central air conditioning (Y=Yes, N=No is the reference category).

1. Extract and display the coefficient, standard error, t-statistic, and p-value for `CentralAir_Y`
2. Write a brief interpretation (1-2 sentences) explaining what this coefficient means in the context of predicting home prices. Write this in a print() statement so that it prints out below after the coefficient, standard error, t-stat, and p-value.

**Checkpoint Question**: What is the coefficient value for `CentralAir_Y`? Round to 2 decimal places.

```python
# Question 14:
feature_name = 'CentralAir_Y'

if feature_name in results.params.index:
    coef = results.params[feature_name]
    std_err = results.bse[feature_name]
    t_stat = results.tvalues[feature_name]
    p_val = results.pvalues[feature_name]

    print(f'{feature_name} coefficient: {coef:.2f}')
    print(f'{feature_name} std error: {std_err:.2f}')
    print(f'{feature_name} t-stat: {t_stat:.2f}')
    print(f'{feature_name} p-value: {p_val:.6f}')

    direction = 'higher' if coef >= 0 else 'lower'
    significance = 'statistically significant' if p_val < 0.05 else 'not statistically significant'

    print(
        f'Interpretation: Holding all other variables constant, homes with central air (Y) are predicted '        f'to sell for ${abs(coef):,.2f} {direction} than comparable homes without central air (N, reference group). '        f'This estimate is {significance} at the 0.05 level.'
    )
else:
    print(f'{feature_name} was not found in the model features.')

```

## Question 15: Interpret a Numeric Feature Coefficient

Interpret the coefficient for the `OverallQual` feature in your model. This variable represents the overall material and finish quality of the home on a scale from 1 (Very Poor) to 10 (Very Excellent).

1. Extract and display the coefficient, standard error, t-statistic, and p-value for `OverallQual`

# Part 5: Optional - Feature Scaling

This section is optional but demonstrates feature scaling concepts from Chapter 9. You can skip this if you're short on time, but it's good practice.

## Question 16 (Optional): Scale Numeric Features

Apply Min-Max scaling to the original numeric features (not the dummy-coded categorical features). Then rebuild your MLR model with the scaled features and compare the results.

**Important**: Only scale the numeric features that existed before dummy coding. Do NOT scale the dummy-coded categorical variables.

Print out the R2 value after scaling rounded to 4 decimal places. 

**Checkpoint Question**: How does this value compare to the R2 of the same model before it was scaled?

```python
# Question 16:
# Scale only the original numeric features (before dummy coding)
X_original = df.drop(columns=['SalePrice']).copy()
original_numeric_cols = X_original.select_dtypes(include=[np.number]).columns.tolist()

X_scaled_base = X_original.copy()
scaler = MinMaxScaler()
X_scaled_base[original_numeric_cols] = scaler.fit_transform(X_scaled_base[original_numeric_cols])

# Dummy-code categorical columns after scaling numeric features
X_scaled = pd.get_dummies(X_scaled_base, drop_first=True)
scaled_non_numeric = X_scaled.select_dtypes(exclude=[np.number]).columns.tolist()
if scaled_non_numeric:
    X_scaled[scaled_non_numeric] = X_scaled[scaled_non_numeric].astype(int)

X_scaled_const = sm.add_constant(X_scaled, has_constant='add')
scaled_results = sm.OLS(y, X_scaled_const).fit()

print(f'R2 before scaling: {results.rsquared:.4f}')
print(f'R2 after scaling: {scaled_results.rsquared:.4f}')

r2_delta = scaled_results.rsquared - results.rsquared
if abs(r2_delta) < 1e-10:
    comparison = 'the same as'
elif r2_delta > 0:
    comparison = 'higher than'
else:
    comparison = 'lower than'

print(f'The scaled-model R2 is {comparison} the unscaled-model R2 by {abs(r2_delta):.4f}.')

```
