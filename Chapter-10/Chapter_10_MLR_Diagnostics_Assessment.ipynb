# Part 1: Load Data and Build Initial Model

In this section, you'll load the dataset, perform basic data cleaning (as learned in prior chapters), and build an initial MLR model that you'll then diagnose in subsequent parts.

## Question 1: Load and Prepare the Dataset

Load the `housing.csv` file and perform basic data cleaning:
1. Remove rows with missing `SalePrice` values
2. Impute missing numeric values with median
3. Impute missing categorical values with mode
4. Separate label (`SalePrice`) and features
5. Dummy code categorical variables using `drop_first=True`
6. Ensure all features are numeric

**Checkpoint Question**: After preparation, how many features (columns) are in your feature matrix `X`? (Exclude the label)

```python
# Question 1:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from pathlib import Path

from scipy.stats import probplot
from sklearn.preprocessing import PowerTransformer
from statsmodels.stats.stattools import omni_normtest
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Load and clean data
data_path = Path('housing.csv')
if not data_path.exists():
    data_path = Path('Chapter-10/housing.csv')
housing = pd.read_csv(data_path)
housing = housing.dropna(subset=['SalePrice']).copy()

numeric_cols = housing.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = housing.select_dtypes(exclude=[np.number]).columns.tolist()

housing[numeric_cols] = housing[numeric_cols].fillna(housing[numeric_cols].median())
for col in categorical_cols:
    mode = housing[col].mode(dropna=True)
    fill_value = mode.iloc[0] if not mode.empty else 'Missing'
    housing[col] = housing[col].fillna(fill_value)

# Label and features
y = housing['SalePrice'].astype(float)
X_raw = housing.drop(columns=['SalePrice'])

# Dummy code categoricals and enforce numeric dtype
X = pd.get_dummies(X_raw, drop_first=True)
X = X.apply(pd.to_numeric, errors='coerce').astype(float)
if X.isna().any().any():
    X = X.fillna(X.median())

print(f'Rows after cleaning: {housing.shape[0]}')
print(f'Checkpoint - Number of features in X: {X.shape[1]}')
```

## Question 2: Build Initial MLR Model

Build a multiple linear regression model using `statsmodels.api.OLS` with all features. Display the model summary.

**Checkpoint Question**: What is the R-squared value for the initial model? Round your answer to 4 decimal places.

```python
# Question 2:
X_const = sm.add_constant(X, has_constant='add')
initial_model = sm.OLS(y, X_const).fit()

print(initial_model.summary())
print(f"Checkpoint - Initial R-squared: {initial_model.rsquared:.4f}")
```

**Checkpoint Question**: What is the skewness of the `SalePrice` distribution? Round your answer to 4 decimal places.

## Question 4: Apply Label Transformations

If the label is skewed, apply transformations to reduce skewness. Try the following transformations:
1. Natural log: `np.log(y)`
2. Square root: `np.sqrt(y)`

Calculate and compare the skewness of each transformed version.

**Checkpoint Question**: What is the skewness of the natural log transformation (`np.log(SalePrice)`)? Round your answer to 4 decimal places.

```python
# Question 4:
y_log = np.log(y)
y_sqrt = np.sqrt(y)

pt = PowerTransformer(method='box-cox', standardize=False)
y_boxcox = pd.Series(
    pt.fit_transform(y.values.reshape(-1, 1)).flatten(),
    index=y.index,
    name='SalePrice_boxcox'
)

transform_skewness = pd.DataFrame({
    'Transformation': ['Original', 'Log', 'Sqrt', 'Box-Cox'],
    'Skewness': [y.skew(), y_log.skew(), y_sqrt.skew(), y_boxcox.skew()]
})

print(transform_skewness)
print(f"Checkpoint - Skewness of log(SalePrice): {y_log.skew():.4f}")
```

## Question 5: Test Residual Normality

Using your initial model (from Question 2), evaluate residual normality:
1. Extract residuals from the model
2. Calculate the Omnibus test statistic and p-value (use `omni_normtest` from `statsmodels.stats.stattools`)
3. Create a histogram of residuals with a KDE overlay

**Note**: A high Omnibus p-value indicates that residual normality is not strongly violated.

**Checkpoint Question**: What is the Omnibus p-value for the initial model's residuals? Round your answer to 6 decimal places.

```python
# Question 5:
residuals_initial = initial_model.resid
omni_stat, omni_pvalue = omni_normtest(residuals_initial)

print(f'Omnibus test statistic: {omni_stat:.4f}')
print(f'Omnibus p-value: {omni_pvalue:.6f}')
print(f'Checkpoint - Omnibus p-value: {omni_pvalue:.6f}')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(residuals_initial, kde=True, bins=30, ax=axes[0])
axes[0].set_title('Initial Model Residuals: Histogram + KDE')
axes[0].set_xlabel('Residual')
axes[0].set_ylabel('Frequency')

probplot(residuals_initial, dist='norm', plot=axes[1])
axes[1].set_title('Initial Model Residuals: Q-Q Plot')

plt.tight_layout()
plt.show()
```

Calculate the Variance Inflation Factor (VIF) for all features in your model:

## Question 11: Apply Polynomial Feature Transformations for Linearity

Polynomial transformations (like squared terms) can capture non-linear relationships and improve model fit. Based on analysis of the refined model, two features show strong non-linear relationships with the log-transformed sale price:
1. **MasVnrArea** (Masonry veneer area) - has a quadratic relationship
2. **OpenPorchSF** (Open porch square feet) - has a quadratic relationship

**Instructions**:
1. Create squared (polynomial) transformations for both features:
   - `MasVnrArea_sq = MasVnrArea ** 2`
   - `OpenPorchSF_sq = OpenPorchSF ** 2`
2. Add these squared terms to your feature set (keep the original features as well)
3. Rebuild the model with the polynomial features (using the log-transformed label from Question 10)
4. Check the statistical significance (p-value) of the squared terms
5. Report the R² and Adjusted R² for the model with polynomial transformations

**Checkpoint Question**: After adding the squared terms for MasVnrArea and OpenPorchSF, what is the Adjusted R-squared value? Round your answer to 4 decimal places.

```python
# Question 11:
X_poly = X_refined.copy()

# Re-add base features if they were removed by VIF filtering in Question 7
for feature in ['MasVnrArea', 'OpenPorchSF']:
    if feature not in X_poly.columns:
        X_poly[feature] = X[feature] if feature in X.columns else 0.0

X_poly['MasVnrArea_sq'] = X_poly['MasVnrArea'] ** 2
X_poly['OpenPorchSF_sq'] = X_poly['OpenPorchSF'] ** 2

X_poly_const = sm.add_constant(X_poly, has_constant='add')
model_poly = sm.OLS(y_log, X_poly_const).fit()

print(model_poly.summary())
print('Squared-term p-values:')
print(model_poly.pvalues[['MasVnrArea_sq', 'OpenPorchSF_sq']])
print(f"Checkpoint - Adjusted R-squared (with polynomial terms): {model_poly.rsquared_adj:.4f}")
```

## Question 12: Create Interaction Effects

Interaction effects capture how the relationship between one feature and the outcome depends on the value of another feature. Based on analysis, the following interaction is theoretically meaningful and statistically significant:
- **MasVnrArea_sq * BsmtQual_Fa**: The effect of masonry veneer area (squared) may depend on basement quality. Homes with lower basement quality (Fa = Fair) may benefit differently from masonry veneer than homes with higher basement quality.

**Instructions**:
1. Create the interaction term: `MasVnrArea_sq * BsmtQual_Fa`
   - Use the squared term from Question 11 (`MasVnrArea_sq`)
   - Multiply it by the basement quality dummy variable (`BsmtQual_Fa`)
2. Add this interaction term to your feature set (keep all existing features including the polynomial terms from Question 11)
3. Rebuild the model with the interaction term (using log-transformed label)
4. Check the statistical significance (p-value) of the interaction term
5. Report the R² and Adjusted R² for the model with the interaction

**Checkpoint Question**: What is the p-value for the `MasVnrArea_sq * BsmtQual_Fa` interaction term? Round your answer to 4 decimal places.

```python
# Question 12:
X_interact = X_poly.copy()

# Ensure the dummy needed for interaction exists
if 'BsmtQual_Fa' not in X_interact.columns:
    X_interact['BsmtQual_Fa'] = X['BsmtQual_Fa'] if 'BsmtQual_Fa' in X.columns else 0.0

X_interact['MasVnrArea_sq_x_BsmtQual_Fa'] = X_interact['MasVnrArea_sq'] * X_interact['BsmtQual_Fa']

X_interact_const = sm.add_constant(X_interact, has_constant='add')
model_interact = sm.OLS(y_log, X_interact_const).fit()

interaction_pvalue = model_interact.pvalues['MasVnrArea_sq_x_BsmtQual_Fa']

print(model_interact.summary())
print(f"Interaction term p-value: {interaction_pvalue:.4f}")
print(f"R-squared: {model_interact.rsquared:.4f}")
print(f"Adjusted R-squared: {model_interact.rsquared_adj:.4f}")
print(f"Checkpoint - Interaction p-value: {interaction_pvalue:.4f}")
```
