```python
import pandas as pd

df = pd.read_csv('insurance.csv')
display(df.head())
```

```python
import statsmodels.api as sm

y = df['charges']
x = df[['age', 'bmi', 'children']]

x = sm.add_constant(x)
display(x.head())

model = sm.OLS(y, x).fit()
display(model.summary())
```

```python
df_dummy = df.copy()

y = df_dummy['charges']
x = df_dummy.drop(columns=['charges']).assign(const=1)
x = pd.get_dummies(x, columns=['region', 'sex', 'smoker'], drop_first=True)
x[x.select_dtypes(bool).columns] = x[x.select_dtypes(bool).columns].astype(int)

model = sm.OLS(y, x).fit()
display(model.summary())

x.head()
```
